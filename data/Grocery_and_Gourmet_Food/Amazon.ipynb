```python
import os
import gzip
import subprocess
import pandas as pd
import numpy as np
from datetime import datetime
```

```python
from tqdm.notebook import tqdm

# ============================================================================
# 配置参数
# ============================================================================
DATASET = 'Grocery_and_Gourmet_Food'
RAW_PATH = os.path.join('./', DATASET)
DATA_FILE = f'reviews_{DATASET}_5.json.gz'
META_FILE = f'meta_{DATASET}.json.gz'

RANDOM_SEED = 0
GDR_PATH = './AmazonGDR/'  # DiffRec数据保存路径

# ============================================================================
# 辅助函数
# ============================================================================
def parse(path):
    """解析gzip压缩的json文件"""
    g = gzip.open(path, 'rb')
    for l in g:
        yield eval(l)

def get_df(path):
    """将gzip文件转换为DataFrame"""
    i = 0
    df = {}
    for d in parse(path):
        df[i] = d
        i += 1
    return pd.DataFrame.from_dict(df, orient='index')

# ============================================================================
# 下载数据（如果不存在）
# ============================================================================
if not os.path.exists(RAW_PATH):
    os.makedirs(RAW_PATH, exist_ok=True)

if not os.path.exists(os.path.join(RAW_PATH, DATA_FILE)):
    print(f'下载交互数据到 {RAW_PATH}')
    subprocess.call(
        f'cd {RAW_PATH} && curl -O http://snap.stanford.edu/data/amazon/productGraph/categoryFiles/reviews_{DATASET}_5.json.gz',
        shell=True
    )

# ============================================================================
# 加载数据
# ============================================================================
print("加载交互数据...")
data_df = get_df(os.path.join(RAW_PATH, DATA_FILE))

# 重命名列并选择需要的列
interaction_df = data_df.rename(columns={
    'asin': 'item_id',
    'reviewerID': 'user_id',
    'unixReviewTime': 'time'
})[['user_id', 'item_id', 'time']]

# 去重并排序
interaction_df = interaction_df.drop_duplicates(['user_id', 'item_id', 'time'])
interaction_df = interaction_df.sort_values(by=['time', 'user_id'], kind='mergesort').reset_index(drop=True)

print(f"原始数据统计: {len(interaction_df)} 条交互, {interaction_df['user_id'].nunique()} 用户, {interaction_df['item_id'].nunique()} 物品")

# ============================================================================
# 5-core过滤（可选，但推荐）
# ============================================================================
def five_core_filter(df, min_interactions=5):
    """过滤掉交互次数少于min_interactions的用户和物品"""
    print("执行5-core过滤...")

    while True:
        # 计算用户和物品的频率
        user_counts = df['user_id'].value_counts()
        item_counts = df['item_id'].value_counts()

        # 筛选出满足条件的用户和物品
        valid_users = set(user_counts[user_counts >= min_interactions].index)
        valid_items = set(item_counts[item_counts >= min_interactions].index)

        print(f"有效用户: {len(valid_users)}/{len(user_counts)}, 有效物品: {len(valid_items)}/{len(item_counts)}")

        # 过滤数据
        filtered_df = df[
            df['user_id'].isin(valid_users) &
            df['item_id'].isin(valid_items)
        ].copy()

        # 如果过滤后数据没有变化，结束循环
        if len(filtered_df) == len(df):
            break

        df = filtered_df

    print(f"5-core过滤后: {len(df)} 条交互, {df['user_id'].nunique()} 用户, {df['item_id'].nunique()} 物品")
    return df

interaction_df = five_core_filter(interaction_df, min_interactions=5)

# ============================================================================
# 重新映射ID为连续整数（从1开始）
# ============================================================================
print("重新映射ID...")
# 用户ID映射
uids = sorted(interaction_df['user_id'].unique())
user2id = dict(zip(uids, range(1, len(uids) + 1)))

# 物品ID映射
iids = sorted(interaction_df['item_id'].unique())
item2id = dict(zip(iids, range(1, len(iids) + 1)))

# 应用映射
interaction_df['user_id'] = interaction_df['user_id'].apply(lambda x: user2id[x])
interaction_df['item_id'] = interaction_df['item_id'].apply(lambda x: item2id[x])

# 保存ID映射（可选）
os.makedirs(GDR_PATH, exist_ok=True)
import json
with open(os.path.join(GDR_PATH, 'user2id.json'), 'w') as f:
    json.dump({str(k): v for k, v in user2id.items()}, f)
with open(os.path.join(GDR_PATH, 'item2id.json'), 'w') as f:
    json.dump({str(k): v for k, v in item2id.items()}, f)

print(f"ID映射完成: {interaction_df['user_id'].nunique()} 用户, {interaction_df['item_id'].nunique()} 物品")

# ============================================================================
# 数据划分：Leave-Last-Out策略
# ============================================================================
print("划分数据集 (Leave-Last-Out)...")

# 按用户和时间排序
interaction_gdr = interaction_df[['user_id', 'item_id', 'time']].copy()
interaction_gdr = interaction_gdr.sort_values(['user_id', 'time'])

# 初始化列表
train_data = []
dev_data = []
test_data = []

# 按用户分组处理
for user_id, group in tqdm(interaction_gdr.groupby('user_id'), desc="处理用户"):
    # 如果交互记录太少（少于3条），全放进训练集
    if len(group) < 3:
        train_data.append(group)
        continue

    # 转换为列表
    items = group['item_id'].tolist()
    times = group['time'].tolist()

    # Test: 最后一个
    test_data.append([user_id, items[-1], times[-1]])

    # Dev: 倒数第二个
    dev_data.append([user_id, items[-2], times[-2]])

    # Train: 剩下的
    train_df_user = pd.DataFrame({
        'user_id': [user_id] * (len(items) - 2),
        'item_id': items[:-2],
        'time': times[:-2]
    })
    train_data.append(train_df_user)

# 合并数据
train_df = pd.concat(train_data, ignore_index=True)
dev_df = pd.DataFrame(dev_data, columns=['user_id', 'item_id', 'time'])
test_df = pd.DataFrame(test_data, columns=['user_id', 'item_id', 'time'])

print(f"Train size: {len(train_df)}")
print(f"Dev size:   {len(dev_df)}")
print(f"Test size:  {len(test_df)}")
print(f"用户数量: {interaction_gdr['user_id'].nunique()}")
print(f"物品数量: {interaction_gdr['item_id'].nunique()}")

# ============================================================================
# 保存数据
# ============================================================================
print(f"保存数据到 {GDR_PATH}...")

train_df.to_csv(os.path.join(GDR_PATH, 'train.csv'), index=False, sep='\t')
dev_df.to_csv(os.path.join(GDR_PATH, 'dev.csv'), index=False, sep='\t')
test_df.to_csv(os.path.join(GDR_PATH, 'test.csv'), index=False, sep='\t')

print("DiffRec数据集准备完成")
```

```python
data_df = get_df(os.path.join(RAW_PATH, DATA_FILE))
data_df.head()
```

```python
meta_df = get_df(os.path.join(RAW_PATH, META_FILE))
meta_df.head()
```

```python
# Only retain items that appear in interaction data

useful_meta_df = meta_df[meta_df['asin'].isin(data_df['asin'])].reset_index(drop=True)
all_items = set(useful_meta_df['asin'].values.tolist())

def related_filter(related_dict):
    out_dict = dict()
    if related_dict is not np.nan:
        for r in related_dict:
            out_dict[r] = list(all_items & set(related_dict[r]))
    return out_dict

useful_meta_df['related'] = useful_meta_df['related'].apply(related_filter)
```

# Build Dataset

### Interaction data

```python
np.random.seed(RANDOM_SEED)
```

```python
out_df = data_df.rename(columns={'asin': 'item_id', 'reviewerID': 'user_id', 'unixReviewTime': 'time'})
out_df = out_df[['user_id', 'item_id', 'time']]
out_df = out_df.drop_duplicates(['user_id', 'item_id', 'time'])
out_df = out_df.sort_values(by=['time', 'user_id'], kind='mergesort').reset_index(drop=True)
out_df.head()
```

```python
# reindex (start from 1)

uids = sorted(out_df['user_id'].unique())
user2id = dict(zip(uids, range(1, len(uids) + 1)))
iids = sorted(out_df['item_id'].unique())
item2id = dict(zip(iids, range(1, len(iids) + 1)))

out_df['user_id'] = out_df['user_id'].apply(lambda x: user2id[x])
out_df['item_id'] = out_df['item_id'].apply(lambda x: item2id[x])
out_df.head()
```

```python
# leave one out spliting

clicked_item_set = dict()
for user_id, seq_df in out_df.groupby('user_id'):
    clicked_item_set[user_id] = set(seq_df['item_id'].values.tolist())
    
def generate_dev_test(data_df):
    result_dfs = []
    n_items = data_df['item_id'].value_counts().size
    for idx in range(2):
        result_df = data_df.groupby('user_id').tail(1).copy()
        data_df = data_df.drop(result_df.index)
        neg_items = np.random.randint(1, n_items + 1, (len(result_df), NEG_ITEMS))
        for i, uid in enumerate(result_df['user_id'].values):
            user_clicked = clicked_item_set[uid]
            for j in range(len(neg_items[i])):
                while neg_items[i][j] in user_clicked:
                    neg_items[i][j] = np.random.randint(1, n_items + 1)
        result_df['neg_items'] = neg_items.tolist()
        result_dfs.append(result_df)
    return result_dfs, data_df
```

```python
leave_df = out_df.groupby('user_id').head(1)
data_df = out_df.drop(leave_df.index)

[test_df, dev_df], data_df = generate_dev_test(data_df)
train_df = pd.concat([leave_df, data_df]).sort_index()

len(train_df), len(dev_df), len(test_df)
```

### Item Metadata

```python
# level-2 category

l2_cate_lst = list()
for cate_lst in useful_meta_df['categories']:
    l2_cate_lst.append(cate_lst[0][2] if len(cate_lst[0]) > 2 else np.nan)
useful_meta_df['l2_category'] = l2_cate_lst  
l2_cates = sorted(useful_meta_df['l2_category'].dropna().unique())
l2_dict = dict(zip(l2_cates, range(1, len(l2_cates) + 1)))
useful_meta_df['l2_category'] = useful_meta_df['l2_category'].apply(lambda x: l2_dict[x] if x == x else 0)
```

```python
item_meta_data = dict()
for idx in range(len(useful_meta_df)):
    info = useful_meta_df.iloc[idx]['related']
    item_meta_data[idx] = {
        'item_id': item2id[useful_meta_df.iloc[idx]['asin']],
        'i_category': useful_meta_df.iloc[idx]['l2_category'],
        'r_complement': list(map(lambda x: item2id[x], info['also_bought'])) if 'also_bought' in info else [],
        'r_substitute': list(map(lambda x: item2id[x], info['also_viewed'])) if 'also_viewed' in info else [],
    }

item_meta_df = pd.DataFrame.from_dict(item_meta_data, orient='index')
item_meta_df = item_meta_df[['item_id', 'i_category', 'r_complement', 'r_substitute']]
item_meta_df.head()
```

```python
# save results

item_meta_df.to_csv(os.path.join(RAW_PATH, 'item_meta.csv'), sep='\t', index=False)
```
